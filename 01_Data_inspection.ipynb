```python
def build_district_dataset(district):

    # Load rainfall
    ppt_data = pd.read_csv(
        os.path.join(dataset_path, "DailyPPT", ppt_file_map[district]),
        header=None,
        skiprows=1,
        names=["Date", "Rainfall"]
    )

    # Load dew point
    dew_data = pd.read_csv(
        os.path.join(dataset_path, "DailyDewTemp", dew_file_map[district]),
        header=None,
        skiprows=1,
        names=["Date", "DewPoint"]
    )

    # Load mean temperature
    temp_data = pd.read_csv(
        os.path.join(dataset_path, "DailyMeanTemp", temp_file_map[district]),
        header=None,
        skiprows=1,
        names=["Date", "MeanTemperature"]
    )

    # Convert dates
    ppt_data["Date"] = pd.to_datetime(ppt_data["Date"], format="%m/%d/%Y")
    dew_data["Date"] = pd.to_datetime(dew_data["Date"], format="%Y-%m-%d")
    temp_data["Date"] = pd.to_datetime(temp_data["Date"], format="%Y-%m-%d")

    # Merge datasets
    daily_data = ppt_data.merge(dew_data, on="Date", how="left")
    daily_data = daily_data.merge(temp_data, on="Date", how="left")

    # Add district
    daily_data["District"] = district

    # Calculate relative humidity
    daily_data["RelativeHumidity"] = 100 * np.exp(
        (17.625 * daily_data["DewPoint"] / (243.04 + daily_data["DewPoint"]))
        -
        (17.625 * daily_data["MeanTemperature"] / (243.04 + daily_data["MeanTemperature"]))
    )

    return daily_data
```

```python
def build_district_dataset(district):

    # Load rainfall
    ppt_data = pd.read_csv(
        os.path.join(dataset_path, "DailyPPT", ppt_file_map[district]),
        header=None,
        skiprows=1,
        names=["Date", "Rainfall"]
    )

    # Load dew point
    dew_data = pd.read_csv(
        os.path.join(dataset_path, "DailyDewTemp", dew_file_map[district]),
        header=None,
        skiprows=1,
        names=["Date", "DewPoint"]
    )

    # Load mean temperature
    temp_data = pd.read_csv(
        os.path.join(dataset_path, "DailyMeanTemp", temp_file_map[district]),
        header=None,
        skiprows=1,
        names=["Date", "MeanTemperature"]
    )

    # Convert dates
    ppt_data["Date"] = pd.to_datetime(ppt_data["Date"], format="mixed")
    dew_data["Date"] = pd.to_datetime(dew_data["Date"], format="%Y-%m-%d")
    temp_data["Date"] = pd.to_datetime(temp_data["Date"], format="%Y-%m-%d")

    # Merge datasets
    daily_data = ppt_data.merge(dew_data, on="Date", how="left")
    daily_data = daily_data.merge(temp_data, on="Date", how="left")

    # Add district
    daily_data["District"] = district

    # Calculate relative humidity
    daily_data["RelativeHumidity"] = 100 * np.exp(
        (17.625 * daily_data["DewPoint"] / (243.04 + daily_data["DewPoint"]))
        -
        (17.625 * daily_data["MeanTemperature"] / (243.04 + daily_data["MeanTemperature"]))
    )

    return daily_data
```

```python
district_boundaries = district_boundaries.to_crs(epsg=32644)

print(district_boundaries.crs)
```

```python
neighbours = {}

for i, district in district_boundaries.iterrows():

    district_name = district["shapeName"]
    neighbouring_districts = []

    for j, other_district in district_boundaries.iterrows():

        if i != j:
            shared_boundary = (
                district.geometry.boundary.intersection(
                    other_district.geometry.boundary
                ).length
            )

            if shared_boundary > 0:
                neighbouring_districts.append(
                    other_district["shapeName"]
                )

    neighbours[district_name] = neighbouring_districts
```

```python
for district, neighbouring_districts in neighbours.items():
    print(district, ":", neighbouring_districts)
```

```python
for i, district in district_boundaries.iterrows():

    for j, other_district in district_boundaries.iterrows():

        if i < j:
            shared_length = (
                district.geometry.boundary.intersection(
                    other_district.geometry.boundary
                ).length
            )

            if 0 < shared_length < 1000:
                print(
                    district["shapeName"],
                    "-",
                    other_district["shapeName"],
                    ":",
                    round(shared_length, 2),
                    "metres"
                )
```

```python
geographical_features = pd.DataFrame({
    "District": list(neighbours.keys()),
    "NeighbouringDistricts": [
        ", ".join(districts)
        for districts in neighbours.values()
    ],
    "NumberOfNeighbours": [
        len(districts)
        for districts in neighbours.values()
    ]
})

geographical_features.head(10)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import geopandas as gpd
```

```python
weekly_climate = pd.read_csv(
    "weekly_climate_weather_geographical.csv"
)
```

```python
weekly_climate.shape
```

```python
weather_features = weekly_climate[
    [
        "Rainfall",
        "MeanTemperature",
        "DewPoint",
        "RelativeHumidity",
        "WeeklyMaxTemperature",
        "WeeklyMinTemperature"
    ]
]
```

```python
weather_features.corr()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    weather_features.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix of Weather Features")
plt.show()
```

```python
weather_features.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

weather_scaled = scaler.fit_transform(weather_features)
```

```python
print(weather_scaled.shape)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()

weather_pca = pca.fit_transform(weather_scaled)
```

```python
print(pca.explained_variance_ratio_)
```

```python
cumulative_variance = np.cumsum(
    pca.explained_variance_ratio_
)

print(cumulative_variance)
```

```python
pca = PCA(n_components=3)

weather_pca = pca.fit_transform(weather_scaled)
```

```python
pca_features = pd.DataFrame(
    weather_pca,
    columns=["PC1", "PC2", "PC3"]
)

pca_features.head()
```

```python
loadings = pd.DataFrame(
    pca.components_.T,
    columns=["PC1", "PC2", "PC3"],
    index=weather_features.columns
)

loadings
```

```python
weekly_climate["PC1"] = pca_features["PC1"]
weekly_climate["PC2"] = pca_features["PC2"]
weekly_climate["PC3"] = pca_features["PC3"]
```

```python
weekly_climate.shape
```

```python
weekly_climate.to_csv(
    "weekly_climate_with_PCA.csv",
    index=False
)
```

```python
weekly_climate.columns.tolist()
```

```python
weekly_climate.isnull().sum()
```

```python
pca_all = PCA()

pca_all.fit(weather_scaled)

explained_variance = pca_all.explained_variance_ratio_ * 100
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    ["PC1", "PC2", "PC3", "PC4", "PC5", "PC6"],
    explained_variance
)

plt.xlabel("Principal Components")
plt.ylabel("Explained Variance (%)")
plt.title("PCA Explained Variance")

plt.show()
```

```python
cumulative = np.cumsum(
    pca_all.explained_variance_ratio_
) * 100

plt.figure(figsize=(9, 5))

plt.plot(
    range(1, 7),
    cumulative,
    marker="o"
)

plt.axhline(
    y=95,
    linestyle="--",
    label="95% variance"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance (%)")
plt.title("PCA Cumulative Explained Variance")

plt.xticks(range(1, 7))
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.heatmap(
    loadings,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("PCA Component Loadings")

plt.show()
```
